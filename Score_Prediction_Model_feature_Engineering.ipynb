```python
final_df.head()
```

```python
X = final_df.drop(columns=['Match_id','Runs_Scored_x','Runs_Scored_y'])
y = final_df['Runs_Scored_y']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=1)
X_test
```

```python
import numpy as np

trf = ColumnTransformer([
    ('trf',OneHotEncoder(sparse_output=False,drop='first'),['Team_Batting','Team_Bowling','Venue_Name','City_Name'])
]
,remainder='passthrough')
```

```python
pipe = Pipeline(steps=[
    ('step1',trf),
    ('step2',StandardScaler()),
    ('step3',XGBRegressor(n_estimators=1000,learning_rate=0.2,max_depth=12,random_state=1))
])
```

```python
pipe.fit(X_train,y_train)
y_pred = pipe.predict(X_test)
print("r2_score:",r2_score(y_test,y_pred))
print("MAE:",mean_absolute_error(y_test,y_pred))
# print("Accuracy:",accuracy_score(y_test,y_pred))
```

Voting model

```python
from sklearn.ensemble import VotingRegressor
from sklearn.linear_model import LinearRegression
from xgboost import XGBRegressor

# Create two models
model1 = XGBRegressor(n_estimators=1000, learning_rate=0.2, max_depth=12, random_state=1)
model2 = LinearRegression()

# Combine them
voting_model = VotingRegressor(estimators=[('xgb', model1), ('lr', model2)], weights=[2, 1])

# Use this voting_model in your pipeline step 3
pipe4 = Pipeline(steps=[
    ('step1', trf),
    ('step2', StandardScaler()),
    ('step3', voting_model)
])

pipe4.fit(X_train, y_train)
Vote_pred=pipe4.predict(X_test)
print("R2 Score:", r2_score(y_test, Vote_pred))
print("MAE:",mean_absolute_error(y_test,Vote_pred))
```

```python
regressors = {
    'Linear Regression': LinearRegression(),
    'Voting Regressor': voting_model,
    'Random Forest': RandomForestRegressor(n_estimators=100, random_state=1),
    'XGBoost': XGBRegressor(n_estimators=1000,learning_rate=0.2,max_depth=12,random_state=1),
    'CatBoost': CatBoostRegressor(verbose=0, random_state=1),
    'LGMBoost': LGBMRegressor(n_estimators=1000, learning_rate=0.2, random_state=1)
}
```

```python
results = []

for name, model in regressors.items():
    # 1. Create a fresh pipeline for each model
    # Note: 'trf' is your existing ColumnTransformer from previous steps
    pipe_model = Pipeline(steps=[
        ('step1', trf),
        ('step2', StandardScaler()), 
        ('step3', model)
    ])
    
    # 2. Train the pipeline
    pipe_model.fit(X_train, y_train)
    
    # 3. Make predictions
    Model_pred = pipe_model.predict(X_test)
    
    # 4. Calculate metrics
    r2 = r2_score(y_test, Model_pred)
    mae = mean_absolute_error(y_test, Model_pred)
    mse4 = mean_squared_error(y_test, Model_pred)

    # 2. Take the square root to get RMSE
    rmse4 = np.sqrt(mse4)
    # 5. Store results
    results.append({
        'Algorithm': name,
        'R2 Score': r2,
        'MAE (Runs)': mae,
        'MSE ': mse4,
        'RMSE': rmse4,
    })
    
    print(f"{name} completed.")

# Convert to DataFrame and sort by best R2 Score
results_df = pd.DataFrame(results).sort_values(by='R2 Score', ascending=False)
```

Match Progression (Dynamic Prediction) This picks a random match from your data
and shows how the prediction changed ball-by-ball.

```python
# 1. Pick a random Match_id from your dataset
random_match_id = final_df['Match_id'].sample(1).values[0]
match_data = final_df[final_df['Match_id'] == random_match_id].sort_values('balls_left', ascending=False)

# 2. Predict scores for this match ball-by-ball
# Note: We need to drop the target and ID columns to match the training input X
X_match = match_data.drop(columns=['Match_id', 'Runs_Scored_x', 'Runs_Scored_y'])
match_preds = pipe.predict(X_match)

# 3. Plot
plt.figure(figsize=(12, 6))
# X-axis is 'balls bowled' (derived from 120 - balls_left)
balls_bowled = 120 - match_data['balls_left']

plt.plot(balls_bowled, match_preds, label='Model Prediction', color='blue', linewidth=2)
plt.axhline(y=match_data['Runs_Scored_y'].iloc[0], color='green', linestyle='--', label=f"Actual Final Score ({match_data['Runs_Scored_y'].iloc[0]})")

plt.title(f'Match Progression: Prediction vs Actual (Match ID: {random_match_id})')
plt.xlabel('Balls Bowled')
plt.ylabel('Predicted Final Score')
plt.legend()
plt.grid(True)
plt.show()
```
